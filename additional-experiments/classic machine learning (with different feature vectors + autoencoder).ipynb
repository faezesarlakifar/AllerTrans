# ProtT5

```python
# @markdown Load Dataset
path = '/content/drive/MyDrive/allergen-detection/embeddings/'

protbert_path = path+'protBERT-embeddings-with-id/'
protbert_train = pd.read_csv(protbert_path+'protBERT_train.csv', index_col=0)
protbert_test = pd.read_csv(protbert_path+'protBERT_test.csv', index_col=0)


df_train = protbert_train
df_test = protbert_test

df_train = df_train.drop('id', axis=1)
df_test = df_test.drop('id', axis=1)

# Convert Label to integer
df_train['Label'] = df_train['Label'].astype(int)
df_test['Label'] = df_test['Label'].astype(int)

# Split data into train and validation sets
X_train = df_train.drop('Label', axis=1)
y_train = df_train['Label']
X_test = df_test.drop('Label', axis=1)
y_test = df_test['Label']

# Create new feature names
feature_names = ['feat_'+str(i) for i in range(X_train.shape[1])]

# Rename columns
X_train.columns = feature_names
X_test.columns = feature_names
```

## Using AutoEncoder

```python
# Convert to numpy arrays
y_train = np.array(y_train)
y_test = np.array(y_test)

# Create Tensors from Labels
y_train = torch.from_numpy(y_train).long()
y_test = torch.from_numpy(y_test).long()
```

```python
# Standardize the data
# scaler = StandardScaler()
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Convert to PyTorch tensors
X_train_scaled = torch.from_numpy(X_train_scaled).float()
X_test_scaled = torch.from_numpy(X_test_scaled).float()
```

```python
X_train = X_train_scaled
X_test = X_test_scaled
```

```python
# Define autoencoder
class Autoencoder(nn.Module):
    def __init__(self, input_dim, encoding_dim):
        super(Autoencoder, self).__init__()

        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.Linear(128, encoding_dim)
        )

        self.decoder = nn.Sequential(
            nn.Linear(encoding_dim, 128),
            nn.ReLU(),
            nn.Linear(128, input_dim),
            nn.Sigmoid()
        )

    def forward(self, x):
        encoded = self.encoder(x)
        decoded = self.decoder(encoded)
        return decoded
```

```python
# Create and train autoencoder
model = Autoencoder(X_train_scaled.shape[1], 32)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001, weight_decay=1e-5)
```

```python
model.train()
for epoch in range(300):
    recon = model(X_train_scaled)
    loss = criterion(recon, X_train_scaled)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    print(f"Epoch {epoch+1} with loss: {loss.item()}")
```

```python
# Encode training and test data
X_train_encoded = model.encoder(X_train_scaled)
X_test_encoded = model.encoder(X_test_scaled)
```

```python
# Train classifiers
rf = RandomForestClassifier(n_estimators=50, min_samples_split=5, max_depth=30, random_state=42)
rf.fit(X_train_encoded.detach().numpy(), y_train.detach().numpy())

svm = SVC(gamma='auto', probability=True)
svm.fit(X_train_encoded.detach().numpy(), y_train.detach().numpy())

knn = KNeighborsClassifier(n_neighbors=10)
knn.fit(X_train_encoded.detach().numpy(), y_train.detach().numpy())

xgb = XGBClassifier(n_estimators=100, learning_rate=0.1)
xgb.fit(X_train_encoded.detach().numpy(), y_train.detach().numpy())
```

## Reported results

```python
X_test_encoded.shape
```

```python
# Evaluate on test set
y_rf_pred = rf.predict(X_test_encoded.detach().numpy())
y_svm_pred = svm.predict(X_test_encoded.detach().numpy())
y_knn_pred = knn.predict(X_test_encoded.detach().numpy())
y_xgb_pred = xgb.predict(X_test_encoded.detach().numpy())
print("Random Forest Test Accuracy after AutoEncoder: ", accuracy_score(y_test, y_rf_pred))
print("SVM Test Accuracy after AutoEncoder: ", accuracy_score(y_test, y_svm_pred))
print("KNN Test Accuracy after AutoEncoder: ", accuracy_score(y_test, y_knn_pred))
print("XGBoost Test Accuracy after AutoEncoder: ", accuracy_score(y_test, y_xgb_pred))
```

```python
print('RF result:\n', classification_report(y_test , y_rf_pred ))
print('svm result:\n', classification_report(y_test , y_svm_pred ))
print('knn result:\n', classification_report(y_test , y_knn_pred ))
print('xgb result:\n', classification_report(y_test , y_xgb_pred ))
```

```python
from sklearn.metrics import (
    accuracy_score, recall_score, confusion_matrix,
    roc_auc_score, matthews_corrcoef
)

def evaluate_model(y_true, model, X_test):

    proba = model.predict_proba(X_test)

    # Get predicted class labels
    pred = model.predict(X_test)

    # Evaluation metrics
    acc = accuracy_score(y_true, pred)
    sensitivity = recall_score(y_true, pred)

    # Confusion Matrix for specificity calculation
    tn, fp, fn, tp = confusion_matrix(y_true, pred).ravel()
    specificity = tn / (tn + fp)

    # AUC
    auc = roc_auc_score(y_true, proba[:, 1])

    # MCC
    mcc = matthews_corrcoef(y_true, pred)

    # Return a dictionary with the results
    return {
        "Accuracy": acc,
        "Sensitivity": sensitivity,
        "Specificity": specificity,
        "AUC": auc,
        "MCC": mcc
    }
```

```python
# Evaluate SVM model
svm_result = evaluate_model(y_test, svm, X_test_encoded.detach().numpy())
print("SVM Evaluation Metrics:")
for metric, value in svm_result.items():
    print(f"{metric}: {value:.4f}")

# Evaluate Random Forest model
rf_result = evaluate_model(y_test, rf, X_test_encoded.detach().numpy())
print("Random Forest Evaluation Metrics:")
for metric, value in rf_result.items():
    print(f"{metric}: {value:.4f}")

# Evaluate KNN model
knn_result = evaluate_model(y_test, knn, X_test_encoded.detach().numpy())
print("KNN Evaluation Metrics:")
for metric, value in knn_result.items():
    print(f"{metric}: {value:.4f}")

# Evaluate XGBoost model
xgb_result = evaluate_model(y_test, xgb, X_test_encoded.detach().numpy())
print("XGBoost Evaluation Metrics:")
for metric, value in xgb_result.items():
    print(f"{metric}: {value:.4f}")
```
